```python
import pandas as pd

df = pd.read_csv('/kaggle/input/datasets/spscientist/students-performance-in-exams/StudentsPerformance.csv')

df.head()
```

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='gender', y='math score')

plt.title('Average Math Score by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Math Score')

plt.show()
```

```python
gender_counts = df['gender'].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Gender Distribution of Students')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['math score'], bins=10, edgecolor='black')

plt.title('Distribution of Maths Scores')
plt.xlabel('Maths Scores')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df[['math score', 'reading score', 'writing score']])

plt.title('Comparison of Student Scores')
plt.xlabel('Subjects')
plt.ylabel('Marks')

plt.show()
```

```python
plt.figure(figsize=(8, 6))

correlation = df[
    ['math score', 'reading score', 'writing score']
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Between Student Scores')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='math score',
    y='reading score',
    hue='gender'
)

plt.title('Maths Score vs Reading Score')
plt.xlabel('Maths Score')
plt.ylabel('Reading Score')

plt.show()
```

# Student Performance Data Visualization

## Project Overview
This project explores student performance using Python, Pandas, Matplotlib, and Seaborn. Six different visualizations were created to understand student scores and their relationships.

## Visualizations
1. Bar Chart – Average Maths Score by Gender.
2. Pie Chart – Gender Distribution.
3. Histogram – Distribution of Maths Scores.
4. Box Plot – Comparison of Maths, Reading, and Writing Scores.
5. Correlation Heatmap – Relationship Between Subject Scores.
6. Scatter Plot – Maths Score vs Reading Score.

## Project Insights
- The bar chart compares average Maths scores by gender.
- The pie chart shows the proportion of male and female students.
- The histogram displays how Maths scores are distributed.
- The box plot compares score distributions across three subjects.
- The heatmap shows positive relationships between subject scores.
- The scatter plot helps identify the relationship between Maths and Reading scores.

## Conclusion
